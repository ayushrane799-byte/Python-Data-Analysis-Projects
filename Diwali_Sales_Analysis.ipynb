Exploratory Data Analysis

Gender


```python
gender_counts =  df['Gender'].value_counts()

# Plotting BAr chart
custom_colors = ['#3498db', '#e74c3c']
plt.bar(gender_counts.index, gender_counts.values,color = custom_colors)

plt.title("Shopping Count by Gender")
plt.ylabel("Number of Shopping Trips")
plt.xlabel("Gender")


plt.show()
```

```python
# Using Seaborn

ax = sns.countplot(x ='Gender', data = df, palette=custom_colors)

for bars in ax.containers:
    ax.bar_label(bars)
```

Age

```python
sns.countplot(x = 'Age Group', data= df,palette=custom_colors)
```

```python
ax = sns.countplot(x = 'Age Group', data= df,palette=custom_colors, hue = 'Gender')

for bars in ax.containers:
    ax.bar_label(bars)
    
```

```python
#  Total Amount  Vs Age Group

sales_age = df.groupby(['Age Group'], as_index= False)['Amount'].sum().sort_values(by ='Amount', ascending = False)
sns.barplot(x= 'Age Group',y = 'Amount', data=sales_age,hue='Age Group', palette='viridis')


# FIX THE Y-AXIS: TTurn off the '1e7' scientific notation
ax.ticklabel_format(style='plain', axis='y')

# Optional: Add comma formatting to make large numbers highly readable (e.g., 40,000,000)
import matplotlib.ticker as ticker
ax.get_yaxis().set_major_formatter(ticker.FuncFormatter(lambda x, p: format(int(x), ',')))
```

Most of the buyers are of age group 26-35 and Females have the purchasing power

State

```python
df.columns
```

```python
State = df.groupby(['State'],as_index= False)['Orders'].sum().sort_values(by = 'Orders',ascending= False).head(10)

sns.set(rc={'figure.figsize':(20,10)})
sns.barplot(data = State , x = 'State', y = 'Orders',palette='viridis',hue ='State',legend= False)
```

```python
# Total Amount / Sales  from top 10 States
Top_State = df.groupby(['State'],as_index= False)['Amount'].sum().sort_values(by = 'Amount',ascending= False).head(10)

sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(data =Top_State  , x = 'State', y = 'Amount',palette='viridis',hue ='State',legend= False)
```

Occupation

```python
sns.set(rc={'figure.figsize': (20,7)})

occupation = sns.countplot(data =df ,x = 'Occupation', palette='viridis')

for bars in occupation.containers:
    occupation.bar_label(bars)
```

```python
Occupation_Spending = df.groupby(['Occupation'],as_index= False)['Amount'].sum().sort_values(by = 'Amount',ascending= False)

sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(data =Occupation_Spending  , x = 'Occupation', y = 'Amount',palette='viridis',hue ='Occupation',legend= False)
```

Most of the buyers are working in IT , Aviation and Health Care

Product Category

```python
df.columns
```

```python
sns.set(rc={'figure.figsize': (25,11)})

product = sns.countplot(data =df ,x = 'Product_Category', palette='viridis')

for bars in product.containers:
    product.bar_label(bars)
```

```python
Product_Spending = df.groupby(['Product_Category'],as_index= False)['Amount'].sum().sort_values(by = 'Amount',ascending= False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data =Product_Spending  , x = 'Product_Category', y = 'Amount',palette='viridis',hue ='Product_Category',legend= False)
```

Most of the sold products are food,Clothing and Electronics

```python
Product_ID = df.groupby(['Product_ID'],as_index= False)['Orders'].sum().sort_values(by = 'Orders',ascending= False).head(10)

sns.set(rc={'figure.figsize':(12,7)})
sns.barplot(data =Product_ID  , x = 'Product_ID', y = 'Orders',palette='viridis',hue ='Product_ID',legend= False)
```

Conclusion

Married woman  from age group 26-35 yrs from UP, Maharashtra and Karnataka working in IT, Healthcare and Aviation are more likely to order from food , clothing and Electonics Category
